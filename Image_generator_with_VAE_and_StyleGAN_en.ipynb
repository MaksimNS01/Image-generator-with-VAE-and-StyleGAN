## Brief description of the problem

Generating realistic images is one of the key tasks in the field of computer vision and machine learning. With the development of deep learning, there has been a need to create algorithms that can not only recognize and classify visual data, but also generate new, previously non-existent images with high quality and diversity.

The problem of image generation involves several aspects:

1. **Creating new content**: The need to generate images that do not exist in the training sample but still look realistic;
2. **Control over characteristics**: The ability to control individual aspects of the generated images (age, gender, emotions, etc.);
3. **Generation diversity**: Ensuring a wide range of different images without repetition;
4. **Image quality**: Generating clear, detailed images without artifacts.

Traditional approaches to image generation based on manual programming or simple statistical models are unable to cope with these tasks. Modern generative models, such as VAE and StyleGAN, offer fundamentally new approaches to solving these problems.

## Key concepts

### Generative models

Generative models are a class of machine learning algorithms that learn to model the probability distribution of data. Unlike discriminative models, which learn to distinguish between classes, generative models learn to create new examples similar to the training data.

**Key characteristics of generative models:**
- Model the joint distribution P(X,Y) or the marginal distribution P(X);
- Can generate new data samples;
- Allow for various types of inference and interpolation.

### Latent space

Latent space (or hidden space) is a multidimensional vector space in which each point represents a compact representation of the original data. In the context of generative models:

- **Encoding**: Converting an image into a latent vector;
- **Decoding**: Converting the latent vector back into an image;
- **Interpolation**: Smooth transition between images through latent space.

### Variational inference

Variational inference is an approximate Bayesian inference method used in VAEs. It transforms the inference problem into an optimization problem, minimizing the discrepancy between the true posterior distribution and the approximate distribution.

### Generative Adversarial Networks (GAN)

GAN is an architecture consisting of two neural networks that learn simultaneously:
- **Generator**: Creates fake data in an attempt to deceive the discriminator;
- **Discriminator**: Attempts to distinguish real data from generated data.

### Reparameterization trick

A technique that allows stochastic variables to be differentiated. Instead of direct sampling from the distribution, a deterministic transformation of the normal distribution is used.

## Mathematical justification

### Variational Autoencoder (VAE)

VAE is based on variational inference and solves the problem of maximizing the lower bound of likelihood (ELBO):

**Objective function:**
$$\mathcal{L}(\theta, \phi; x) = \mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x|z)] - D_{KL}(q_\phi(z|x) || p(z))$$

**Components:**
- $\mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x|z)]$ - expected likelihood (recovery loss)
- $D_{KL}(q_\phi(z|x) || p(z))$ - KL divergence between the posterior and prior distributions

**KL divergence for normal distributions:**
$$D_{KL} = \frac{1}{2} \sum_{i=1}^{d} (\mu_i^2 + \sigma_i^2 - \log \sigma_i^2 - 1)$$

**Reparameterization trick:**
$$z = \mu + \sigma \odot \epsilon, \quad \epsilon \sim \mathcal{N}(0, I)$$

### Generative adversarial networks (GAN)

GANs are based on a minimax game between a generator and a discriminator:

**Objective function:**
$$\min_G \max_D V(D, G) = \mathbb{E}_{x \sim p_{data}(x)}[\log D(x)] + \mathbb{E}_{z \sim p_z(z)}[\log(1 - D(G(z)))]$$

**Discriminator loss function:**
$$\mathcal{L}_D = -\mathbb{E}_{x \sim p_{data}}[\log D(x)] - \mathbb{E}_{z \sim p_z}[\log(1 - D(G(z)))]$$

**Generator loss function:**
$$\mathcal{L}_G = -\mathbb{E}_{z \sim p_z}[\log D(G(z))]$$

### AdaIN (Adaptive Instance Normalization)

StyleGAN uses AdaIN for feature stylization:

$$AdaIN(x_i, w) = y_{s,i} \cdot \frac{x_i - \mu(x_i)} {\sigma(x_i)} + y_{b,i}$$

Where:
- $x_i$ - feature map
- $\mu(x_i)$ and $\sigma(x_i)$ - mean and standard deviation
- $y_{s,i}$ and $y_{b,i}$ - scaling and shift parameters from the style vector

### Quality metrics

**Mean square error (MSE):**
$$MSE = \frac{1}{n} \sum_{i=1}^{n} (x_i - \hat{x}_i)^2$$

**Mean absolute error (MAE):**
$$MAE = \frac{1}{n} \sum_{i=1}^{n} |x_i - \hat{x}_i|$$

**KL divergence:**
$$D_{KL}(P||Q) = \sum_i P(i) \log \frac{P(i)}{Q(i)}$$

```python
#@title Preprocessing { display-mode: "form" }
# Path to the folder with images
data_dir = '/content/dataset/img_align_celeba'

# Get a list of paths to all .jpg files in the specified directory
image_paths = [os.path.join(data_dir, f) for f in os.listdir(data_dir) if f.endswith(‘.jpg’)]

def load_and_preprocess_image(image_path):
    # Load image
    image = tf.io.read_file(image_path)

    # Decode PNG image with 3 channels (RGB)
    image = tf.image.decode_png(image, channels=3)

    # Convert data type to float32 for further processing
    image = tf.cast(image, tf.float32)

    # Resize (usually 64x64 or 128x128 for CelebA)
    image = tf.image.resize(image, [64, 64])

    # Normalize to the range [-1, 1] or [0, 1]
    image = (image / 127.5) - 1.0  # for [-1, 1]
    # Or: image = image / 255.0   # for [0, 1]

    return image

# Creating a dataset from paths
def create_dataset(image_paths, batch_size=32, shuffle=True):
    # Creating a TensorFlow dataset from a list of image paths
    dataset = tf.data.Dataset.from_tensor_slices(image_paths)
    # Apply the preprocessing function to each image in parallel
    dataset = dataset.map(load_and_preprocess_image, num_parallel_calls=tf.data.AUTOTUNE)

    # Shuffle the data, if enabled
    if shuffle:
        dataset = dataset.shuffle(buffer_size=1000)

    # Group images into batches of a specified size
    dataset = dataset.batch(batch_size)
    # Preload the next batch while training the current one (performance optimization)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)

    return dataset
```

# 5. IMPLEMENTATION OF THE VAE MODEL

## Defining VAE Architecture

## Theoretical Foundations of Variational Autoencoders

A variational autoencoder (VAE) is a generative model that combines the ideas of traditional autoencoders and Bayesian inference. Unlike conventional autoencoders, VAE not only learns to compress and reconstruct data, but also models the probability distribution in the latent space.

### Mathematical basis

VAE is based on variational inference and solves the problem of maximizing the evidence lower bound (ELBO):

$$\mathcal{L}(\theta, \phi; x) = \mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x|z)] - D_{KL}(q_\phi(z|x) || p(z))$$

Where:
- $x$ — input data
- $z$ — latent variables
- $q_\phi(z|x)$ — approximate posterior distribution (encoder)
- $p_\theta(x|z)$ — data likelihood (decoder)
- $p(z)$ — prior distribution (usually standard normal)

### VAE architecture

The VAE architecture consists of two main components:

#### Encoder
The encoder takes the input image $x$ and converts it into latent distribution parameters:
- **Input**: Image $x \in \mathbb{R}^{H \times W \times C}$
- **Intermediate layers**: Convolution layers with nonlinearities for feature extraction
- **Output**: Parameters of the normal distribution $\mu \in \mathbb{R}^d$ and $\log \sigma^2 \in \mathbb{R}^d$


#### Decoder
The decoder takes the latent vector and reconstructs the image:
- **Input**: Latent vector $z \in \mathbb{R}^d$
- **Intermediate layers**: Transposed convolution layers for image reconstruction
- **Output**: Reconstructed image $\hat{x}$

### Training process

1. **Forward pass**:
   - The encoder converts the input image $x$ into parameters $\mu$ and $\log \sigma^2$
   - A reparameterization trick is applied: $z = \mu + \sigma \odot \epsilon$, where $\epsilon \sim \mathcal{N}(0, I)$
   - The decoder converts $z$ into the reconstructed image $\hat{x}$

2. **Loss calculation**:
   - Reconstruction loss: $\mathcal{L}_{recon} = ||x - \hat{x}||^2$
   - KL divergence: $\mathcal{L}_{KL} = \frac{1}{2} \sum_{i=1}^{d} (\mu_i^2 + \sigma_i^2 - \log \sigma_i^2 - 1)$
   - Total loss: $\mathcal{L} = \mathcal{L}_{recon} + \mathcal{L}_{KL}$

### Advantages of VAE

1. **Normalized latent space**: Thanks to KL regularization, latent vectors follow a normal distribution;
2. **Interpretability**: The influence of different dimensions of the latent space can be analyzed;
3. **Stable training**: Fewer vanishing mode issues compared to GAN;
4. **Likelihood estimation**: Ability to evaluate model quality through data likelihood.

### Disadvantages of VAE

1. **Blurry images**: Due to the minimization of the mean square error, images may be blurry;
2. **Limited expressiveness**: Complex data distributions may be poorly modeled;
3. **Posterior collapse**: In some cases, the model may ignore latent variables.

```python
#@title VAE hyperparameters tuning { display-mode: "form" }
#@markdown Training Parameters:
epochs = 50 # @param {"type":"number", min:1, max:1000}
learning_rate = 0.0002  #@param {type:"number", min:1e-10, max:0.1}
latent_dim = 128  #@param {type:"integer", min:1, max:4096}
```

```python
#@title Implementation of VAE { display-mode: "form" }
class VAE(tf.keras.Model):
    """Variational autocoder"""

    def __init__(self, latent_dim):
        super(VAE, self).__init__()
        self.latent_dim = latent_dim

        # Encoder - convolutional network for encoding images into latent space
        self.encoder = tf.keras.Sequential([
            tf.keras.layers.InputLayer(input_shape=(64, 64, 3)),
            # First convolutional layer: 32 filters, 3x3 kernel, stride 2, padding='same'
            tf.keras.layers.Conv2D(32, 3, strides=2, padding='same', activation='relu'),
            # Batch normalization for training stability
            tf.keras.layers.BatchNormalization(),
            # Second convolutional layer: 64 filters, 3x3 kernel, stride 2, padding='same'
            tf.keras.layers.Conv2D(64, 3, strides=2, padding='same', activation='relu'),
            # Batch normalization
            tf.keras.layers.BatchNormalization(),
            # Third convolutional layer: 128 filters, 3x3 kernel, stride 2, padding='same'
            tf.keras.layers.Conv2D(128, 3, strides=2, padding='same', activation='relu'),
            # Batch normalization
            tf.keras.layers.BatchNormalization(),
            # Conversion to a one-dimensional vector
            tf.keras.layers.Flatten(),
            # Use a linear layer to obtain distribution parameters
            tf.keras.layers.Dense(latent_dim + latent_dim),  # [mean, logvar]
        ])

        # Decoder - transposed convolutional network for decoding from latent space to image
        self.decoder = tf.keras.Sequential([
            tf.keras.layers.InputLayer(input_shape=(latent_dim,)),
            # Fully connected layer for restoring features before image reconstruction
            tf.keras.layers.Dense(8*8*128, activation='relu'),
            # Converting a one-dimensional vector back to a 3D tensor
            tf.keras.layers.Reshape(target_shape=(8, 8, 128)),
            # First transposed convolution layer: 128 filters, 3x3 kernel, stride 2
            tf.keras.layers.Conv2DTranspose(128, 3, strides=2, padding='same', activation='relu'),
            # Batch normalization
            tf.keras.layers.BatchNormalization(),
            # Second transposed convolution layer: 64 filters, 3x3 kernel, stride 2
            tf.keras.layers.Conv2DTranspose(64, 3, strides=2, padding='same', activation='relu'),
            # Batch normalization
            tf.keras.layers.BatchNormalization(),
            # Third transposed convolution layer: 32 filters, 3x3 kernel, stride 2
            tf.keras.layers.Conv2DTranspose(32, 3, strides=2, padding='same', activation='relu'),
            # Batch normalization
            tf.keras.layers.BatchNormalization(),
            # Final transposed convolution layer: 3 channels (RGB), tanh activation for range [-1, 1]
            tf.keras.layers.Conv2DTranspose(3, 3, strides=1, padding='same', activation='tanh'),  # tanh for [-1, 1]
        ])

    def encode(self, x):
        """Encoding an image into latent space"""
        # Pass the image through the encoder and split it into the mean and log variance
        mean, logvar = tf.split(self.encoder(x), num_or_size_splits=2, axis=1)
        return mean, logvar

    def reparameterize(self, mean, logvar):
        """Reparameterization for training"""
        # Generate random noise with normal distribution
        eps = tf.random.normal(shape=mean.shape)
        # Apply reparameterization trick: z = μ + ε * σ
        return eps * tf.exp(logvar * .5) + mean

    def decode(self, z):
        """Decode from latent space to image"""
        return self.decoder(z)

    def call(self, x):
        """Full pass through VAE"""
        # Encoding the input image
        mean, logvar = self.encode(x)
        # Applying reparameterization to obtain the latent vector
        z = self.reparameterize(mean, logvar)
        # Decoding the latent vector into a reconstructed image
        x_recon = self.decode(z)
        return x_recon, mean, logvar

# Creating a VAE instance with a specified latent space dimension
vae = VAE(latent_dim)
```

```python
#@title Loss functions and metrics for VAE { display-mode: "form" }
@tf.function
def compute_loss(model, x):
    """Calculation of the VAE loss function"""
    # Direct pass through the model: obtain reconstruction and latent distribution parameters
    x_recon, mean, logvar = model(x)

    # MSE reconstruction loss - measures the difference between the original and reconstructed images
    recon_loss = tf.reduce_mean(tf.square(x - x_recon))

    # KL divergence - regularizes the latent space, forcing it to be close to the standard normal distribution
    kl_loss = -0.5 * tf.reduce_mean(1 + logvar - tf.square(mean) - tf.exp(logvar))

    # Total loss - sum of the reconstruction loss function and KL divergence
    total_loss = recon_loss + kl_loss

    return total_loss, recon_loss, kl_loss

@tf.function
def train_step(model, x, optimizer):
    """One training step"""
    # Create GradientTape for automatic differentiation
    with tf.GradientTape() as tape:
        # Compute losses for the current batch
        total_loss, recon_loss, kl_loss = compute_loss(model, x)

    # Calculate loss gradients for trainable model parameters
    gradients = tape.gradient(total_loss, model.trainable_variables)
    # Apply gradients to update model weights
    optimizer.apply_gradients(zip(gradients, model.trainable_variables))

    return total_loss, recon_loss, kl_loss

@tf.function
def val_step(model, x):
    """One validation step"""
    # Calculating losses on a validation batch without calculating gradients
    total_loss, recon_loss, kl_loss = compute_loss(model, x)
    return total_loss, recon_loss, kl_loss
```

# 6. VAE TRAINING

```python
#@title VAE training process with validation { display-mode: "form" }
# Оптимизатор Adam с заданной скоростью обучения
optimizer = tf.keras.optimizers.Adam(learning_rate)

# Metrics for tracking losses during training and validation
train_loss = tf.keras.metrics.Mean()  # Average total loss on the training sample
recon_loss_metric = tf.keras.metrics.Mean()  # Average recovery loss on the training sample
kl_loss_metric = tf.keras.metrics.Mean()  # Average KL divergence on the training sample
val_loss = tf.keras.metrics.Mean()  # Average total loss on the validation sample

# Training history - dictionary for storing metric values by epoch
history = {
    'train_total_loss': [],
    'train_recon_loss': [],
    'train_kl_loss': [],
    'val_total_loss': [],
    'val_recon_loss': [],
    'val_kl_loss': []
}

# Progress bar for visualizing the learning process
pbar = tqdm(total=epochs, desc="VAE training")

# Main training loop
for epoch in range(epochs):
    # Reset training metrics at the beginning of each epoch
    train_loss.reset_state()
    recon_loss_metric.reset_state()
    kl_loss_metric.reset_state()

    # Training on batches of the training sample
    for batch in train_dataset:
        # One training step: calculating losses and updating weights
        total_loss, recon_loss, kl_loss = train_step(vae, batch, optimizer)
        # Updating metrics with current loss values
        train_loss.update_state(total_loss)
        recon_loss_metric.update_state(recon_loss)
        kl_loss_metric.update_state(kl_loss)

    # Validation on the validation sample
    val_loss.reset_state()
    # Creating separate metrics for validation in the current epoch
    val_recon_loss = tf.keras.metrics.Mean()
    val_kl_loss = tf.keras.metrics.Mean()

    # Calculate losses on validation batches
    for batch in val_dataset:
        total_loss, recon_loss, kl_loss = val_step(vae, batch)
        # Update validation metrics
        val_loss.update_state(total_loss)
        val_recon_loss.update_state(recon_loss)
        val_kl_loss.update_state(kl_loss)

    # Save current epoch metrics to training history
    history['train_total_loss'].append(train_loss.result().numpy())
    history['train_recon_loss'].append(recon_loss_metric.result().numpy())
    history['train_kl_loss'].append(kl_loss_metric.result().numpy())
    history['val_total_loss'].append(val_loss.result().numpy())
    history['val_recon_loss'].append(val_recon_loss.result().numpy())
    history['val_kl_loss'].append(val_kl_loss.result().numpy())

    # Update the progress bar with current metric values
    pbar.set_postfix({
        'Train Loss': f'{train_loss.result():.4f}',
        'Val Loss': f'{val_loss.result():.4f}',
        'Recon Loss': f'{recon_loss_metric.result():.4f}',
        'KL Loss': f'{kl_loss_metric.result():.4f}'
    })
    pbar.update(1)

# Closing the progress bar after completing training
pbar.close()

print("\nVAE training has been completed!")
```

```python
#@title Interpolation in latent space VAE { display-mode: "form" }
def interpolate(model, num_interpolations=10):
    """Interpolation between two points in latent space"""
    # Generate two random points in latent space
    z1 = tf.random.normal(shape=(1, model.latent_dim))
    z2 = tf.random.normal(shape=(1, model.latent_dim))

    # Interpolate between them using linear interpolation
    alphas = np.linspace(0, 1, num_interpolations)  # Create uniformly distributed values from 0 to 1
    interpolated_images = []

    # Generate intermediate images
    for alpha in alphas:
        # Linear interpolation between two latent vectors: z = (1-α) * z1 + α * z2
        z_interp = (1 - alpha) * z1 + alpha * z2
        # Decoding the intermediate latent vector into an image
        img = model.decode(z_interp)
        # Adding the decoded image to the list (taking the first element of the batch)
        interpolated_images.append(img[0])

    # Displaying interpolated images
    plt.figure(figsize=(20, 4))
    for i, img in enumerate(interpolated_images):
        # Creating a subplot for each interpolated image
        plt.subplot(1, num_interpolations, i+1)
        # Denormalize the image from the range [-1, 1] to [0, 1] for correct display
        img = (img + 1.0) / 2.0
        plt.imshow(img)
        plt.axis(‘off’)
        # Add a title with the current value of the interpolation parameter α
        plt.title(f'α={alphas[i]:.1f}')
    # Automatic indentation settings
    plt.tight_layout()
    # Displaying the graph
    plt.show()

# Performing interpolation between two random points in the latent space of the VAE
interpolate(vae)
```

```python
#@title Visualization of latent space VAE { display-mode: "form" }
def visualize_latent_space(model, dataset, num_samples=500):
    """Visualization of latent space using t-SNE"""
    # Obtaining encodings for image sampling
    latent_vectors = []  # List for storing latent vectors
    sample_images = []   # List for storing corresponding images

    count = 0
    # Iteration over the dataset to collect latent vectors
    for batch in dataset:
        if count >= num_samples:
            break
        # Determining the size of the current batch, taking into account the remaining number of required samples
        batch_size = min(batch.shape[0], num_samples - count)
        images = batch[:batch_size]
        # Encoding images into latent space (obtaining average values)
        mean, _ = model.encode(images)
        # Saving latent vectors and images
        latent_vectors.append(mean.numpy())
        sample_images.append(images.numpy())
        count += batch_size

    # Combining all batches into a single array
    latent_vectors = np.concatenate(latent_vectors, axis=0)
    sample_images = np.concatenate(sample_images, axis=0)

    # Apply t-SNE to visualize the multidimensional latent space in 2D
    tsne = TSNE(n_components=2, random_state=42, perplexity=30)
    latent_2d = tsne.fit_transform(latent_vectors[:300])  # Limit for speed

    # Creating a graph to visualize the latent space
    plt.figure(figsize=(10, 8))
    # Displaying points in the latent space with transparency
    plt.scatter(latent_2d[:, 0], latent_2d[:, 1], alpha=0.6)
    plt.title("Latent space VAE (t-SNE)")
    plt.xlabel("t-SNE 1")
    plt.ylabel("t-SNE 2")
    plt.show()

    return latent_vectors, sample_images

# Call the function to visualize the latent space of the trained VAE
latent_vectors, sample_images = visualize_latent_space(vae, train_dataset)
```

```python
#@title Calculation of VAE quality metrics { display-mode: "form" }
def calculate_vae_metrics(model, test_dataset, num_samples=1000):
    """
    Calculating various quality metrics for VAE

    Args:
        model: Trained VAE model
        test_dataset: Test dataset
        num_samples: Number of samples for analysis

    Returns:
        Dictionary with quality metrics
    """
    mse_losses = []  # List for storing MSE losses
    mae_losses = []  # List for storing MAE losses
    kl_divergences = []  # List for storing KL divergences
    total_losses = []  # List for storing total losses

    count = 0
    # Iterate over the test dataset to calculate metrics
    for batch in test_dataset:
        if count >= num_samples:
            break

        # Get the original images and their reconstructions through the model
        original_images = batch
        reconstructed_images, mean, logvar = model(original_images)

        # MSE (Mean Squared Error) - mean squared error of reconstruction
        mse = tf.reduce_mean(tf.square(original_images - reconstructed_images), axis=[1, 2, 3])
        mse_losses.extend(mse.numpy())

        # MAE (Mean Absolute Error) - mean absolute error of reconstruction
        mae = tf.reduce_mean(tf.abs(original_images - reconstructed_images), axis=[1, 2, 3])
        mae_losses.extend(mae.numpy())

        # KL divergence - latent space regularization
        kl = -0.5 * tf.reduce_mean(1 + logvar - tf.square(mean) - tf.exp(logvar), axis=1)
        kl_divergences.extend(kl.numpy())

        # Total loss - sum of MSE and KL divergence
        total = mse + kl
        total_losses.extend(total.numpy())

        count += len(batch)

    # Average metrics across all samples
    metrics = {
        'MSE': np.mean(mse_losses),
        'MAE': np.mean(mae_losses),
        'KL_Divergence': np.mean(kl_divergences),
        'Total_Loss': np.mean(total_losses),
        'MSE_Std': np.std(mse_losses),
        'MAE_Std': np.std(mae_losses),
        'KL_Std': np.std(kl_divergences)
    }

    return metrics

print("Calculation of VAE quality metrics...")
vae_metrics = calculate_vae_metrics(vae, test_dataset)

print("\n=== VAE Quality Metrics ===")
for metric_name, value in vae_metrics.items():
    print(f"{metric_name}: {value:.6f}")

# Creating a DataFrame for convenient display
metrics_df = pd.DataFrame([vae_metrics])
print("\nMetrics in tabular form:")
print(metrics_df.to_string(index=False))

# Visualization of metrics
plt.figure(figsize=(12, 6))

# Chart of key quality metrics
plt.subplot(1, 2, 1)
metric_names = ['MSE', 'MAE', 'KL_Divergence', 'Total_Loss']
metric_values = [vae_metrics[name] for name in metric_names]
bars = plt.bar(metric_names, metric_values, color=['blue', 'green', 'red', 'orange'])
plt.title('VAE quality metrics')
plt.ylabel('Value')
plt.xticks(rotation=45)

# Adding numerical values to columns for better readability
for bar, value in zip(bars, metric_values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + value*0.01,
             f'{value:.4f}', ha='center', va='bottom')

# Standard deviation chart for metrics
plt.subplot(1, 2, 2)
std_names = ['MSE_Std', 'MAE_Std', 'KL_Std']
std_values = [vae_metrics[name] for name in std_names]
bars = plt.bar(std_names, std_values, color=['lightblue', 'lightgreen', 'lightcoral'])
plt.title('Standard deviations of metrics')
plt.ylabel('Value')
plt.xticks(rotation=45)

# Adding numerical values to standard deviation columns
for bar, value in zip(bars, std_values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + value*0.01,
             f'{value:.4f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
#@title VAE error analysis { display-mode: "form" }
def analyze_vae_errors(model, test_dataset, num_samples=500):
    """
    VAE error analysis: identifying images with the highest and lowest errors

    Args:
        model: Trained VAE model
        test_dataset: Test dataset
        num_samples: Number of samples for analysis

    Returns:
        Dictionary with error analysis results
    """
    all_errors = []  # List for storing recovery errors for each image
    all_images = []  # List for storing original images
    all_reconstructions = []  # List for storing reconstructed images

    count = 0
    # Iterate over the test dataset to analyze errors
    for batch in test_dataset:
        if count >= num_samples:
            break

        original_images = batch
        # Obtain reconstruction and latent distribution parameters
        reconstructed_images, mean, logvar = model(original_images)

        # Calculate MSE errors for each image in the batch
        errors = tf.reduce_mean(tf.square(original_images - reconstructed_images), axis=[1, 2, 3])

        all_errors.extend(errors.numpy())
        all_images.append(original_images.numpy())
        all_reconstructions.append(reconstructed_images.numpy())

        count += len(batch)

    # Combining all data from different batches into a single array
    all_images = np.concatenate(all_images, axis=0)
    all_reconstructions = np.concatenate(all_reconstructions, axis=0)

    # Sort indices by error magnitude to identify the best and worst cases
    error_indices = np.argsort(all_errors)

    # Obtain image indices with the smallest (best) and largest (worst) errors
    best_indices = error_indices[:8]
    worst_indices = error_indices[-8:]

    # Calculate the mean error and standard deviation
    mean_error = np.mean(all_errors)
    std_error = np.std(all_errors)

    return {
        'all_errors': all_errors,
        'best_indices': best_indices,
        'worst_indices': worst_indices,
        'all_images': all_images,
        'all_reconstructions': all_reconstructions,
        'mean_error': mean_error,
        'std_error': std_error
    }

print("VAE error analysis...")
error_analysis = analyze_vae_errors(vae, test_dataset)

print(f"Average recovery error: {error_analysis['mean_error']:.6f}")
print(f"Standard deviation of the error: {error_analysis['std_error']:.6f}")
print(f"Minimum error: {np.min(error_analysis['all_errors']):.6f}")
print(f"Maximum error: {np.max(error_analysis['all_errors']):.6f}")

# Visualization of error distribution
plt.figure(figsize=(15, 5))

# Histogram of recovery error distribution
plt.subplot(1, 3, 1)
plt.hist(error_analysis['all_errors'], bins=50, alpha=0.7, color='blue')
plt.axvline(error_analysis['mean_error'], color='red', linestyle='--',
            label=f'Среднее: {error_analysis["mean_error"]:.4f}')
plt.axvline(error_analysis['mean_error'] + error_analysis['std_error'],
            color='orange', linestyle='--', alpha=0.7, label='±1 std')
plt.axvline(error_analysis['mean_error'] - error_analysis['std_error'],
            color='orange', linestyle='--', alpha=0.7)
plt.xlabel('Recovery error (MSE)')
plt.ylabel('Frequency')
plt.title('Distribution of recovery errors')
plt.legend()

# Q-Q plot for checking the normality of error distribution
plt.subplot(1, 3, 2)
from scipy import stats
stats.probplot(error_analysis['all_errors'], dist="norm", plot=plt)
plt.title('Q-Q plot errors')
plt.xlabel('Theoretical quantiles')
plt.ylabel('Sample quantiles')

# Box plot of errors for visualizing outliers and quartiles
plt.subplot(1, 3, 3)
plt.boxplot(error_analysis['all_errors'], vert=True)
plt.ylabel('Recovery error')
plt.title('Box plot of recovery errors')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Visualization of the best and worst reconstructions
fig, axes = plt.subplots(4, 8, figsize=(20, 10))

# Best reconstructions (smallest errors) - top two rows
for i, idx in enumerate(error_analysis['best_indices']):
    # Original image
    axes[0, i].imshow((error_analysis['all_images'][idx] + 1.0) / 2.0)
    axes[0, i].set_title(f'Original #{idx}\nError: {error_analysis["all_errors"][idx]:.4f}')
    axes[0, i].axis('off')

    # Reconstruction
    axes[1, i].imshow((error_analysis['all_reconstructions'][idx] + 1.0) / 2.0)
    axes[1, i].set_title(f'Reconstruction #{idx}')
    axes[1, i].axis('off')

# Worst renovations (biggest mistakes) - bottom two lines
for i, idx in enumerate(error_analysis['worst_indices']):
    # Original image
    axes[2, i].imshow((error_analysis['all_images'][idx] + 1.0) / 2.0)
    axes[2, i].set_title(f'Original #{idx}\nError: {error_analysis["all_errors"][idx]:.4f}')
    axes[2, i].axis('off')

    # Reconstruction
    axes[3, i].imshow((error_analysis['all_reconstructions'][idx] + 1.0) / 2.0)
    axes[3, i].set_title(f'Reconstruction #{idx}')
    axes[3, i].axis('off')

plt.suptitle('Analyzing the quality of reconstruction: Best and Worst Examples')
plt.tight_layout()
plt.show()

print("\n=== Error statistics ===")
print(f"25th percentile: {np.percentile(error_analysis['all_errors'], 25):.6f}")
print(f"Median: {np.median(error_analysis['all_errors']):.6f}")
print(f"75th percentile: {np.percentile(error_analysis['all_errors'], 75):.6f}")
print(f"90th percentile: {np.percentile(error_analysis['all_errors'], 90):.6f}")
print(f"95th percentile: {np.percentile(error_analysis['all_errors'], 95):.6f}")

# Outlier detection (errors > Q3 + 1.5 * IQR) - standard method for detecting outliers
Q1 = np.percentile(error_analysis['all_errors'], 25)
Q3 = np.percentile(error_analysis['all_errors'], 75)
IQR = Q3 - Q1
outliers = [err for err in error_analysis['all_errors'] if err > (Q3 + 1.5 * IQR)]
print(f"Number of emissions: {len(outliers)} ({len(outliers)/len(error_analysis['all_errors'])*100:.2f}%)")
```

# 8. IMPLEMENTATION OF THE STYLEGAN MODEL

## Definition of StyleGAN architecture

## Theoretical foundations of StyleGAN

StyleGAN (Style-Based Generator Architecture for Generative Adversarial Networks) is a revolutionary architecture for generative adversarial networks. The main solution is to separate stylistic characteristics from other image factors through adaptive feature normalization (AdaIN).

### Mathematical basis

StyleGAN is based on the classic GAN architecture, but with an improved generator:

$$G: z \rightarrow w \rightarrow \{w_1, w_2, ..., w_L\} \rightarrow \{y_1, y_2, ..., y_L\} \rightarrow x$$

Where:
- $z$ — input latent vector from normal distribution
- $w$ — intermediate representation after mapping network
- $w_i$ — style vectors for each layer
- $y_i$ — feature maps after stylization
- $x$ — generated image

### StyleGAN Architecture

The StyleGAN architecture consists of three main components:

#### Mapping Network
The mapping network converts the input latent vector $z$ into an intermediate representation $w$:
- **Input**: $z \in \mathbb{R}^{512}$ (usually from $\mathcal{N}(0, I)$)
- **Architecture**: Multi-layer perceptron (MLP) with 8 layers
- **Output**: $w \in \mathbb{R}^{512}$

#### Synthesis Network
Synthesis network generates an image from style vectors:
- **Input**: Style vectors $\{w_1, w_2, ..., w_L\}$
- **Architecture**: Cascade of transposed convolutional layers with AdaIN
- **Output**: Generated image $x$

#### AdaIN (Adaptive Instance Normalization)
AdaIN is applied to each layer to stylize features:
$$AdaIN(x_i, w) = y_{s,i} \cdot \frac{x_i - \mu(x_i)} {\sigma(x_i)} + y_{b,i}$$

Where $y_{s,i}$ and $y_{b,i}$ are the scale and shift obtained from the style vector $w$.

### Generation process

1. **Latent vector generation**: $z \sim \mathcal{N}(0, I)$
2. **Mapping**: $w = MappingNetwork(z)$
3. **Stylization**: For each layer $i$:
- Obtaining stylization parameters: $(y_{s,i}, y_{b,i}) = Affine(w)$
   - Apply AdaIN: $y_i = AdaIN(x_i, y_{s,i}, y_{b,i})$
4. **Image synthesis**: Sequential application of transposed convolutional layers

### Key innovations of StyleGAN

#### 1. Progressive Growing
Gradual increase in image resolution during training:
- Starts with small images (4×4);
- Layers are gradually added to increase resolution;
- Improves training stability and result quality.

#### 2. Style Mixing
Mixing styles from different latent vectors:
- Allows you to control different aspects of the image;
- Separation of global and local style characteristics.

#### 3. Truncation Trick
Improving the quality of generated images:
- Limits latent vectors near the center of the distribution;
- Compromise between quality and diversity.

### Advantages of StyleGAN

1. **High image quality**: Generates photorealistic images;
2. **Style control**: Ability to precisely control style characteristics;
3. **Interpretability of latent space**: Well-structured latent space;
4. **Separation of characteristics**: Ability to control individual aspects of the image.

### Disadvantages of StyleGAN

1. **Training complexity**: Requires careful tuning of hyperparameters;
2. **Computational complexity**: Requires significant computational resources;
3. **Mode Dropping**: Risk of ignoring some data modes;
4. **Data-intensive**: Requires large amounts of high-quality data.

```python
#@title StyleGAN hyperparameter tuning{ display-mode: "form" }
#@markdown Training parameters:
gan_epochs = 50 # @param {"type":"number", min:1, max:1000}
gan_learning_rate = 0.0002  #@param {type:"number", min:1e-10, max:0.1}
BATCH_SIZE = 64 # @param {"type":"slider","min":2,"max":1024,"step":1}
```

```python
#@title StyleGAN implementation { display-mode: "form" }
class Generator(tf.keras.Model):
    """StyleGAN generator"""

    def __init__(self, latent_dim):
        super(Generator, self).__init__()
        self.latent_dim = latent_dim

        # Fully connected layer for converting the latent vector into features
        self.dense = tf.keras.layers.Dense(8*8*256, activation='relu')
        # Conversion of a one-dimensional vector into a 3D tensor for further processing
        self.reshape = tf.keras.layers.Reshape((8, 8, 256))

        # First layer of transposed convolution to increase image size
        self.upsample1 = tf.keras.layers.Conv2DTranspose(128, 4, strides=2, padding='same', activation='relu')
        # Batch normalization for training stability
        self.bn1 = tf.keras.layers.BatchNormalization()

        # Second layer of transposed convolution
        self.upsample2 = tf.keras.layers.Conv2DTranspose(64, 4, strides=2, padding='same', activation='relu')
        # Batch normalization
        self.bn2 = tf.keras.layers.BatchNormalization()

        # Third layer of transposed convolution
        self.upsample3 = tf.keras.layers.Conv2DTranspose(32, 4, strides=2, padding='same', activation='relu')
        # Batch normalization
        self.bn3 = tf.keras.layers.BatchNormalization()

        # Final convolution layer to obtain an RGB image with a range of [-1, 1]
        self.output_layer = tf.keras.layers.Conv2D(3, 3, padding='same', activation='tanh')

    def call(self, inputs):
        # Pass through a fully connected layer and change the shape
        x = self.dense(inputs)
        x = self.reshape(x)

        # Sequential application of transposed convolution layers
        x = self.upsample1(x)
        x = self.bn1(x)

        x = self.upsample2(x)
        x = self.bn2(x)

        x = self.upsample3(x)
        x = self.bn3(x)

        # Final top layer
        return self.output_layer(x)

class Discriminator(tf.keras.Model):
    """StyleGAN discriminator (simplified version)"""
    def __init__(self):
        super(Discriminator, self).__init__()

        # First convolutional layer for feature extraction, reduces image size
        self.conv1 = tf.keras.layers.Conv2D(32, 4, strides=2, padding='same', activation='relu')
        # Second convolution layer
        self.conv2 = tf.keras.layers.Conv2D(64, 4, strides=2, padding='same', activation='relu')
        # Batch normalization
        self.bn2 = tf.keras.layers.BatchNormalization()
        # Third convolution layer
        self.conv3 = tf.keras.layers.Conv2D(128, 4, strides=2, padding='same', activation='relu')
        # Batch normalization
        self.bn3 = tf.keras.layers.BatchNormalization()
        # Converting a 3D tensor to a one-dimensional vector
        self.flatten = tf.keras.layers.Flatten()
        # Dropout layer for regularization and preventing overfitting
        self.dropout = tf.keras.layers.Dropout(0.3)
        # Final fully connected layer for binary classification (real/fake)
        self.dense = tf.keras.layers.Dense(1)

    def call(self, inputs):
        # Sequential pass through convolution layers
        x = self.conv1(inputs)
        x = self.conv2(x)
        x = self.bn2(x)
        x = self.conv3(x)
        x = self.bn3(x)
        # Conversion to a one-dimensional vector and application of dropout
        x = self.flatten(x)
        x = self.dropout(x)
        # Final classification
        return self.dense(x)

# Creating instances of generator and discriminator models
generator = Generator(latent_dim)
discriminator = Discriminator()

# Binary cross-entropy loss function with logits
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

def discriminator_loss(real_output, fake_output):
    # Loss for real images (should be classified as 1)
    real_loss = cross_entropy(tf.ones_like(real_output), real_output)
    # Loss for fake images (should be classified as 0)
    fake_loss = cross_entropy(tf.zeros_like(fake_output), fake_output)
    # Total loss for the discriminator
    total_loss = real_loss + fake_loss
    return total_loss

def generator_loss(fake_output):
    # Generator loss - aims to make the discriminator classify fakes as real (1)
    return cross_entropy(tf.ones_like(fake_output), fake_output)
```

# 9. STYLEGAN TRAINING

```python
#@title StyleGAN training process with validation { display-mode: "form" }
# Create Adam optimizers for the generator and discriminator with a specified learning rate
generator_optimizer = tf.keras.optimizers.Adam(gan_learning_rate)
discriminator_optimizer = tf.keras.optimizers.Adam(gan_learning_rate)

@tf.function
def train_step_gan(images):
    # Generate random noise for the generator input
    noise = tf.random.normal([BATCH_SIZE, latent_dim])

    # Create two GradientTapes to calculate gradients simultaneously
    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        # Generate images from noise
        generated_images = generator(noise, training=True)

        # Obtain discriminator outputs for real and generated images
        real_output = discriminator(images, training=True)
        fake_output = discriminator(generated_images, training=True)

        # Calculating losses for the generator and discriminator
        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)

    # Calculating loss gradients for the trainable parameters of the corresponding models
    gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
    gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)

    # Applying gradients to update the weights of the generator and discriminator
    generator_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
    discriminator_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))

    return gen_loss, disc_loss

# GAN training history - dictionary for storing loss values by epoch
gan_history = {
    'gen_loss': [],  # Generator loss
    'disc_loss': []  # Discriminator loss
}

# Progress bar for visualizing the training process
pbar = tqdm(total=gan_epochs, desc="StyleGAN training")

# Main GAN training loop
for epoch in range(gan_epochs):
    gen_losses = []  # List for storing generator losses in the current epoch
    disc_losses = []  # List for storing discriminator losses in the current epoch

    # Training on all batches of the training sample
    for batch in train_dataset:
        # One GAN training step
        gen_loss, disc_loss = train_step_gan(batch)
        # Saving the current loss values
        gen_losses.append(gen_loss)
        disc_losses.append(disc_loss)

    # Saving the averaged metrics of the current epoch to the training history
    gan_history['gen_loss'].append(np.mean(gen_losses))
    gan_history['disc_loss'].append(np.mean(disc_losses))

    # Updating the progress bar with the current average loss values
    pbar.set_postfix({
        'Gen Loss': f'{np.mean(gen_losses):.4f}',
        'Disc Loss': f'{np.mean(disc_losses):.4f}'
    })
    pbar.update(1)

# Closing the progress bar after training is complete
pbar.close()

print("StyleGAN training is completed!")
```

```python
#@title StyleGAN quality analysis { display-mode: "form" }
def analyze_stylegan_quality(generator, discriminator, test_dataset, latent_dim, num_samples=500):
    """
    StyleGAN quality analysis: evaluating the realism and diversity of generated images

    Args:
        generator: Trained StyleGAN generator
        discriminator: Trained StyleGAN discriminator
        test_dataset: Test dataset (real images)
        latent_dim: Latent space dimension
        num_samples: Number of samples for analysis

    Returns:
        Dictionary with quality analysis results
    """
    # Image generation
    noise = tf.random.normal([num_samples, latent_dim])
    generated_images = generator(noise, training=False)

    # Discriminator evaluation
    real_scores = []
    fake_scores = []

    # Evaluation on real images
    count = 0
    for batch in test_dataset:
        if count >= num_samples:
            break
        scores = discriminator(batch, training=False)
        real_scores.extend(scores.numpy().flatten())
        count += len(batch)

    # Evaluation on generated images
    fake_scores = discriminator(generated_images, training=False).numpy().flatten()

    # Diversity analysis (pairwise distances between images)
    sample_size = min(100, num_samples)
    sample_images = generated_images[:sample_size]

    # Calculating the average pairwise distance (limited sample for performance)
    pairwise_distances = []
    for i in range(min(20, sample_size)):
        for j in range(i+1, min(25, sample_size)):
            distance = tf.reduce_mean(tf.square(sample_images[i] - sample_images[j]))
            pairwise_distances.append(distance.numpy())

    diversity_score = np.mean(pairwise_distances) if pairwise_distances else 0

    # Pixel quality analysis
    pixel_mean = tf.reduce_mean(generated_images)
    pixel_std = tf.math.reduce_std(generated_images)

    # Saturation (contrast) analysis
    contrast_measure = tf.reduce_mean(tf.square(generated_images - tf.reduce_mean(generated_images, axis=[1,2,3], keepdims=True)))

    return {
        'real_scores': real_scores,
        'fake_scores': fake_scores,
        'generated_images': generated_images.numpy(),
        'mean_real_score': np.mean(real_scores),
        'mean_fake_score': np.mean(fake_scores),
        'std_real_score': np.std(real_scores),
        'std_fake_score': np.std(fake_scores),
        'diversity_score': diversity_score,
        'pixel_mean': pixel_mean.numpy(),
        'pixel_std': pixel_std.numpy(),
        'contrast_measure': contrast_measure.numpy()
    }

# StyleGAN quality analysis
print("StyleGAN quality analysis...")
quality_analysis = analyze_stylegan_quality(generator, discriminator, test_dataset, latent_dim)

print(f"Average score of real images: {quality_analysis['mean_real_score']:.6f}")
print(f"Average score of generated images: {quality_analysis['mean_fake_score']:.6f}")
print(f"A variety of generation: {quality_analysis['diversity_score']:.6f}")
print(f"The average pixel value: {quality_analysis['pixel_mean']:.6f}")
print(f"Contrast: {quality_analysis['contrast_measure']:.6f}")

# Visualization of discriminator score distribution
plt.figure(figsize=(15, 5))

# Histogram of discriminator scores for real and generated images
plt.subplot(1, 3, 1)
plt.hist(quality_analysis['real_scores'], bins=50, alpha=0.7, label='Real', color='blue')
plt.hist(quality_analysis['fake_scores'], bins=50, alpha=0.7, label='Generated', color='red')
plt.axvline(quality_analysis['mean_real_score'], color='blue', linestyle='--',
            label=f'Average real: {quality_analysis["mean_real_score"]:.4f}')
plt.axvline(quality_analysis['mean_fake_score'], color='red', linestyle='--',
            label=f'Average generated: {quality_analysis["mean_fake_score"]:.4f}')
plt.xlabel('Discriminator rating')
plt.ylabel('Frequency')
plt.title('Distribution of discriminator assessments')
plt.legend()

# Q-Q plot for comparing distributions
plt.subplot(1, 3, 2)
# Comparison of distributions of real and generated estimates
stats.probplot(quality_analysis['real_scores'], dist="norm", plot=plt, fit=True)
plt.title('Q-Q Plot Real estimates')
plt.xlabel('Theoretical quantili')
plt.ylabel('Selective quantili')

# Box plot for comparing distributions
plt.subplot(1, 3, 3)
box_data = [quality_analysis['real_scores'], quality_analysis['fake_scores']]
plt.boxplot(box_data, labels=['Real', 'Generated'])
plt.ylabel('Discriminator score')
plt.title('Comparison of score distributions')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Visualization of examples of generated images with the best and worst scores
# Select images with the highest and lowest discriminator scores
fake_scores_array = np.array(quality_analysis['fake_scores'])
best_indices = np.argsort(fake_scores_array)[-8:]  # 8 best (highest scores)
worst_indices = np.argsort(fake_scores_array)[:8]  # 8 worst (lowest scores)

fig, axes = plt.subplots(4, 8, figsize=(20, 10))

# Best generated images (highest discriminator scores)
for i, idx in enumerate(best_indices):
    # Generated image
    img = (quality_analysis['generated_images'][idx] + 1.0) / 2.0
    axes[0, i].imshow(img)
    axes[0, i].set_title(f'Best #{idx}\nScore: {quality_analysis["fake_scores"][idx]:.4f}')
    axes[0, i].axis('off')

    # Repeated display for visual consistency
    axes[1, i].imshow(img)
    axes[1, i].set_title(f'High quality')
    axes[1, i].axis('off')

# Worst generated images (lowest discriminator scores)
for i, idx in enumerate(worst_indices):
    # Generated image
    img = (quality_analysis['generated_images'][idx] + 1.0) / 2.0
    axes[2, i].imshow(img)
    axes[2, i].set_title(f'Worst #{idx}\nScore: {quality_analysis["fake_scores"][idx]:.4f}')
    axes[2, i].axis('off')

    # Repeated display for visual consistency
    axes[3, i].imshow(img)
    axes[3, i].set_title(f'Low quality')
    axes[3, i].axis('off')

plt.suptitle('Analysis of the quality of generation: Best and worst examples')
plt.tight_layout()
plt.show()

# Quality statistics
print("\n=== Quality statistics ===")
print(f"25th percentile (real): {np.percentile(quality_analysis['real_scores'], 25):.6f}")
print(f"Median (real): {np.median(quality_analysis['real_scores']):.6f}")
print(f"75th percentile (real): {np.percentile(quality_analysis['real_scores'], 75):.6f}")
print(f"25th percentile (generated): {np.percentile(quality_analysis['fake_scores'], 25):.6f}")
print(f"Median (generated): {np.median(quality_analysis['fake_scores']):.6f}")
print(f"75th percentile (generated): {np.percentile(quality_analysis['fake_scores'], 75):.6f}")

# Analysis of the gap between distributions
score_difference = quality_analysis['mean_real_score'] - quality_analysis['mean_fake_score']
print(f"\nGap between average grades: {score_difference:.6f}")
print(f"Standard deviation (real): {quality_analysis['std_real_score']:.6f}")
print(f"Standard deviation (generated): {quality_analysis['std_fake_score']:.6f}")

# Diversity assessment
print(f"\n=== Metrics of diversity ===")
print(f"The average pair distance: {quality_analysis['diversity_score']:.6f}")
print(f"The average pixel value: {quality_analysis['pixel_mean']:.6f}")
print(f"Standard pixel deviation: {quality_analysis['pixel_std']:.6f}")
print(f"The measure of contrast: {quality_analysis['contrast_measure']:.6f}")
```

```python
#@title Interpolation in StyleGAN { display-mode: "form" }
def interpolate_gan(generator, num_interpolations=10):
    """Interpolation between two points in the latent space of StyleGAN"""
    # Generate two random points in latent space
    z1 = tf.random.normal(shape=(1, latent_dim))
    z2 = tf.random.normal(shape=(1, latent_dim))

    # Interpolate between them using linear interpolation
    alphas = np.linspace(0, 1, num_interpolations)  # Create uniformly distributed values from 0 to 1
    interpolated_images = []

    # Generate intermediate images
    for alpha in alphas:
        # Linear interpolation between two latent vectors: z = (1-α) * z1 + α * z2
        z_interp = (1 - alpha) * z1 + alpha * z2
        # Generate an image from the intermediate latent vector (training=False for inference)
        img = generator(z_interp, training=False)
        # Adding the generated image to the list (taking the first element of the batch)
        interpolated_images.append(img[0])

    # Displaying the interpolated images
    plt.figure(figsize=(20, 4))
    for i, img in enumerate(interpolated_images):
        # Creating a subplot for each interpolated image
        plt.subplot(1, num_interpolations, i+1)
        # Denormalizing the image from the range [-1, 1] to [0, 1] for correct display
        img = (img + 1.0) / 2.0
        plt.imshow(img)
        plt.axis('off')
        # Add a title with the current value of the interpolation parameter α
        plt.title(f'α={alphas[i]:.1f}')
    # Automatic indentation adjustment
    plt.tight_layout()
    # Displaying the graph
    plt.show()

# Performing interpolation between two random points in the StyleGAN latent space
interpolate_gan(generator)
```

## Problems and limitations

### Limited availability of datasets
One of the key challenges when working with generative models is the availability of high-quality datasets. The following difficulties were identified during the project implementation:

1. **Problems with CelebA**: Despite its popularity, the CelebA dataset became unavailable through standard TensorFlow Datasets channels due to changes in Google Drive's access policy. This led to the need to search for alternative data sources.

2. **Data fragmentation**: Many large datasets (such as CelebA) are split into parts, which complicates the download process and requires additional steps to assemble and verify data integrity.

### Limitations of cloud platforms
1. **Limited GPU resources**: Free GPUs in Google Colab have restrictions on usage time and memory, which limits the complexity of models and batch sizes.

2. **Execution time restrictions**: Google Colab sessions automatically terminate after a certain period of inactivity, requiring periodic restarting of long training processes.

3. **Storage limitations**: Free accounts have limited disk space, making it difficult to work with large datasets.

### Scaling issues
- **GPU memory**: When working with high-resolution images or large batches, there may be a shortage of video memory.
- **Training time**: Training complex generative models requires a significant amount of time, which may exceed the limitations of cloud platforms.

### Limitations of VAE
1. **Image blurriness**: Due to the minimization of mean squared error, VAEs tend to generate blurry images, especially when the latent space capacity is limited.

2. **Posterior collapse**: In some cases, the model may ignore latent variables, using only the decoder to generate images.

3. **Limited expressiveness**: Simple VAE architectures may lack the power to model complex data distributions.

4. **Balance between loss components**: Careful tuning is required to balance the reconstruction loss and KL divergence.

### Limitations of StyleGAN
1. **Training complexity**: StyleGAN requires careful tuning of hyperparameters and can be unstable during training.

2. **Mode dropping**: The generator may ignore some data modes, resulting in a reduction in the diversity of generated images.

3. **Computational complexity**: High computational resource requirements limit the use of full-scale architectures.

4. **Data requirements**: A large amount of high-quality data is required to achieve high quality.

### Assessment of generation quality
1. **Lack of universal metrics**: Existing metrics (FID, IS) do not always correlate with subjective perception of image quality.

2. **Subjectivity of assessment**: The quality of generated images is difficult to assess automatically and often requires visual verification.

3. **Limited test data**: The size and quality of the test sample affect the reliability of the evaluation results.

### Reproducibility issues
1. **Dependence on random initial conditions**: Results can vary significantly depending on weight initialization.

2. **Dependence on hardware**: Differences in hardware configuration can affect training results.

3. **Version compatibility**: Changes in library versions can affect the reproducibility of results.

### Simplification of architectures
1. **Reduced image sizes**: To speed up training and fit within memory constraints, 64×64 images had to be used instead of the original sizes.

2. **Simplified architectures**: The implemented models are simplified versions of the original architectures to ensure operability under constraints.

3. **Limited number of epochs**: Due to time constraints, the number of training epochs was limited, which undoubtedly affected the quality of the results.

### Optimization issues
1. **Limited configuration options**: In a cloud environment, not all optimization parameters can be configured optimally.

2. **Batch size limitations**: Batch size is limited by available memory, which affects training stability.

3. **Limited monitoring capabilities**: In a cloud environment, access to detailed information about the training process is limited.
